## A. Word Rotation Problem

```python
# Function that moves the first character to the end of the string
def rotate_word(text):
    # Get all characters except the first one
    remaining = text[1:]
    
    # Add the first character to the end
    return remaining + text[0]
```

```python
# Test cases
print(rotate_word("python"))
print(rotate_word("logic"))
print(rotate_word("Code"))
print(rotate_word("A"))
```

## B. Username Builder Problem

```python
# Function that creates a username from the first and last name
def make_username(first_name, last_name):
    # Convert the first name to lowercase and remove spaces
    first_name = first_name.lower().replace(" ", "")
    
    # Convert the last name to lowercase and remove spaces
    last_name = last_name.lower().replace(" ", "")
    
    # Join the first name and last name with a period
    return first_name + "." + last_name
```

```python
# Test cases
print(make_username("Ada", "Lovelace"))
print(make_username("Alan", "Turing"))
print(make_username("Ana Maria", "De Leon"))
```

## C. Bookend Swap Problem

```python
# Function that swaps the first and last elements of a list
def swap_bookends(items):
    # Unpack the first element, middle elements, and last element
    first, *middle, last = items
    
    # Return a new list with the first and last elements swapped
    return [last] + middle + [first]
```

```python
# Test cases
print(swap_bookends([1, 2, 3, 4, 5, 6]))
print(swap_bookends(["red", "green", "blue"]))
print(swap_bookends([8, 3]))
```
